# Analisis Exploratorio de Datos (EDA) - Statlog German Credit Data

Este notebook realiza la exploracion del conjunto de datos de riesgo crediticio German Credit Data (Statlog).
El objetivo es comprender la estructura de las variables, inspeccionar sus distribuciones y analizar las relaciones/correlaciones con la variable objetivo de riesgo crediticio.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from ucimlrepo import fetch_ucirepo

# Configuracion de estilo para los graficos
sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

# Carga del dataset desde UCI Machine Learning Repository
statlog_german_credit = fetch_ucirepo(id=144)

X = statlog_german_credit.data.features.copy()
y = statlog_german_credit.data.targets.copy()

# Diccionario de renombramiento explicito de variables segun el diccionario de datos
column_mapping = {
    'Attribute1': 'status_checking_account',
    'Attribute2': 'duration_months',
    'Attribute3': 'credit_history',
    'Attribute4': 'purpose',
    'Attribute5': 'credit_amount',
    'Attribute6': 'savings_account_bonds',
    'Attribute7': 'present_employment_since',
    'Attribute8': 'installment_rate',
    'Attribute9': 'personal_status_sex',
    'Attribute10': 'other_debtors_guarantors',
    'Attribute11': 'present_residence_since',
    'Attribute12': 'property',
    'Attribute13': 'age_years',
    'Attribute14': 'other_installment_plans',
    'Attribute15': 'housing',
    'Attribute16': 'number_existing_credits',
    'Attribute17': 'job',
    'Attribute18': 'number_people_liable',
    'Attribute19': 'telephone',
    'Attribute20': 'foreign_worker'
}

X = X.rename(columns=column_mapping)
y = y.rename(columns={'class': 'credit_risk'})

# Consolidacion en un unico DataFrame para EDA
df = pd.concat([X, y], axis=1)

# Mapeo de etiqueta objetivo para interpretabilidad grafica (1=Buen Credito, 2=Mal Credito)
df['credit_risk_label'] = df['credit_risk'].map({1: 'Buen Credito', 2: 'Mal Credito'})

print('Primeros 5 registros del dataset:')
df.head()
```

## 1. Estructura y Resumen del Dataset

A continuacion se examinan los tipos de datos, presencia de nulos y estadisticos descriptivos de las variables numericas y categoricas.


```python
print('Informacion general de columnas y tipos de datos:')
df.info()

print('\nEstadisticas descriptivas de variables numericas:')
display(df.describe())

print('\nDistribucion de la variable objetivo (credit_risk):')
display(df['credit_risk_label'].value_counts(normalize=False))
display(df['credit_risk_label'].value_counts(normalize=True) * 100)
```

## 2. Distribucion de Variables Numericas

Se grafican los histogramas y diagramas de caja para analizar el comportamiento de las variables continuas y discretas respecto al riesgo crediticio.


```python
num_cols = ['duration_months', 'credit_amount', 'age_years', 'installment_rate']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for idx, col in enumerate(num_cols):
    sns.histplot(data=df, x=col, hue='credit_risk_label', kde=True, ax=axes[idx], palette={'Buen Credito': '#2ecc71', 'Mal Credito': '#e74c3c'})
    axes[idx].set_title(f'Distribucion de {col}', fontsize=12)
    axes[idx].set_xlabel(col)
    axes[idx].set_ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for idx, col in enumerate(['duration_months', 'credit_amount', 'age_years']):
    sns.boxplot(data=df, x='credit_risk_label', y=col, ax=axes[idx], palette={'Buen Credito': '#2ecc71', 'Mal Credito': '#e74c3c'})
    axes[idx].set_title(f'{col} segun Riesgo Crediticio', fontsize=12)
    axes[idx].set_xlabel('Riesgo Crediticio')
    axes[idx].set_ylabel(col)

plt.tight_layout()
plt.show()
```

## 3. Distribucion de Variables Categoricas Principales

Inspeccion de las frecuencias de atributos categoricos clave segun la condicion de riesgo crediticio.


```python
cat_cols_main = ['status_checking_account', 'credit_history', 'savings_account_bonds', 'present_employment_since']

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
axes = axes.flatten()

for idx, col in enumerate(cat_cols_main):
    sns.countplot(data=df, x=col, hue='credit_risk_label', ax=axes[idx], palette={'Buen Credito': '#2ecc71', 'Mal Credito': '#e74c3c'})
    axes[idx].set_title(f'Frecuencias de {col} por Riesgo Crediticio', fontsize=12)
    axes[idx].set_xlabel(col)
    axes[idx].set_ylabel('Conteo')
    axes[idx].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

## 4. Analisis de Correlaciones

Evaluacion de la correlacion entre variables numericas y el riesgo crediticio (mapeado numericamente: 0 para Buen Credito, 1 para Mal Credito).


```python
df_corr = df.copy()
df_corr['credit_risk_target'] = (df_corr['credit_risk'] == 2).astype(int)

numeric_df = df_corr.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Matriz de Correlacion de Variables Numericas y Objetivo', fontsize=14)
plt.show()

print('Correlacion con la variable objetivo (Mal Credito = 1):')
print(corr_matrix['credit_risk_target'].sort_values(ascending=False))
```
